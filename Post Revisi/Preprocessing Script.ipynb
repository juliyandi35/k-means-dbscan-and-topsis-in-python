# Preprocessing Data
## Data POI

```python
import pandas as pd
```

```python
# Import Data
POI = pd.read_excel("D:/Kerjaan/Project P64180125 (k-Means, DBSCAN and TOPSIS in Python)/Real Data Used/Raw Data/Data-POI.xlsx")
POI.head()
```

```python
from collections import defaultdict

counts = defaultdict(lambda: defaultdict(int))
for kabkot, subclass in zip(POI['KABKOT'], POI['SUBCLASS']):
    counts[kabkot][subclass] += 1

rows = []
for kabkot, subclass_counts in counts.items():
    for subclass, count in subclass_counts.items():
        rows.append({'KABKOT': kabkot, 'SUBCLASS': subclass, 'COUNT': count})

POI_df = pd.DataFrame(rows)
POI_df.head()
```

```python
POI_pivot = POI_df.pivot(index='KABKOT', columns='SUBCLASS', values='COUNT').fillna(0).astype(int)
POI_pivot.head()
```

```python
# Reset index untuk membuat KABKOT menjadi kolom
POI_pivot = POI_pivot.reset_index()

# Mengganti nama kolom 'index' (jika ada) menjadi 'KABKOT'
POI_pivot = POI_pivot.rename(columns={'index': 'KABKOT'})

POI_pivot.head()
```

```python
# Mengganti nilai di kolom 'KABKOT'
POI_pivot['KABKOT'] = POI_pivot['KABKOT'].replace({
    'KOTA BANJAR BARU': 'KOTA BANJARBARU',
    'KOTA BARU': 'KOTABARU'
})
```

```python
POI_pivot.to_excel('D:/Kerjaan/Project P64180125 (k-Means, DBSCAN and TOPSIS in Python)/Real Data Used/Processed Data/Data POI.xlsx')
```

## Jumlah Akomodasi Hotel

```python
Hotel = pd.read_excel("D:/Kerjaan/Project P64180125 (k-Means, DBSCAN and TOPSIS in Python)/Real Data Used/Raw Data/Akomodasi Hotel.xlsx")
Hotel.head()
```

```python
# Mengubah huruf di kolom 'Lokasi/Tahun' menjadi uppercase
Hotel['Lokasi/Tahun'] = Hotel['Lokasi/Tahun'].str.upper()

# Membuat DataFrame baru dengan data tahun 2022
Hotel_2022 = Hotel[['Lokasi/Tahun', 'Hotel (2022)', 'Kamar (2022)', 'Tempat Tidur (2022)']]

# Menghapus "(2022)" dari nama kolom
Hotel_2022 = Hotel_2022.rename(columns={
    'Hotel (2022)': 'Hotel',
    'Kamar (2022)': 'Kamar',
    'Tempat Tidur (2022)': 'Tempat Tidur'
})

# Mengganti nilai di kolom 'Lokasi/Tahun'
Hotel_2022['Lokasi/Tahun'] = Hotel_2022['Lokasi/Tahun'].replace({
    'BANJARMASIN': 'KOTA BANJARMASIN',
    'BANJARBARU': 'KOTA BANJARBARU'
})

# Mengganti "-" dengan 0
Hotel_2022 = Hotel_2022.replace('-', 0)

# Mengubah tipe data kolom menjadi integer
for col in ['Hotel', 'Kamar', 'Tempat Tidur']:
    Hotel_2022[col] = Hotel_2022[col].astype(int)

# Mengubah nama kolom
Hotel_2022['KABKOT'] = Hotel_2022['Lokasi/Tahun']
Hotel_2022 = Hotel_2022.drop(columns=['Lokasi/Tahun'])

# Mengambil kolom KABKOT
kabkot_column = Hotel_2022['KABKOT']

# Menghapus kolom KABKOT dari DataFrame asli
Hotel_2022 = Hotel_2022.drop(columns=['KABKOT'])

# Menambahkan kolom KABKOT ke posisi paling kiri
Hotel_2022.insert(0, 'KABKOT', kabkot_column)

Hotel_2022.head()
```

```python
Hotel_2022.to_excel('D:/Kerjaan/Project P64180125 (k-Means, DBSCAN and TOPSIS in Python)/Real Data Used/Processed Data/Akomodasi Hotel.xlsx')
```
